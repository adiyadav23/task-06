```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv("healthcare_dataset.csv")
```

```python
df.head()
df.info()
df.describe()
```

```python
sns.histplot(df['Age'], bins=20, kde=True, color='skyblue')
plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(data=df, x='Gender', palette='Set2')
plt.title("Count of Patients by Gender")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='Treatment Type', palette='Set3')
plt.title("Count of Each Treatment Type")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=df, x='Diagnosis', y='Age', estimator='mean', palette='coolwarm')
plt.title("Average Age per Diagnosis")
plt.xticks(rotation=45)
plt.show()
```

```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Date of Discharge'] = pd.to_datetime(df['Date of Discharge'])
df['Length of Stay'] = (df['Date of Discharge'] - df['Date of Admission']).dt.days

sns.histplot(df['Length of Stay'], bins=15, kde=True, color='purple')
plt.title("Distribution of Hospital Stay Duration")
plt.xlabel("Number of Days")
plt.ylabel("Count")
plt.show()
```

```python
df['Blood Pressure'].unique()
```

```python
df[['Systolic', 'Diastolic']] = df['Blood Pressure'].str.split('/', expand=True)
df['Systolic'] = pd.to_numeric(df['Systolic'], errors='coerce')
df['Diastolic'] = pd.to_numeric(df['Diastolic'], errors='coerce')
```

```python
sns.histplot(df['Systolic'], bins=20, kde=True, color='orange')
plt.title("Distribution of Systolic Blood Pressure")
plt.xlabel("Systolic (mmHg)")
plt.ylabel("Count")
plt.show()
```

```python
sns.histplot(df['Diastolic'], bins=20, kde=True, color='green')
plt.title("Distribution of Diastolic Blood Pressure")
plt.xlabel("Diastolic (mmHg)")
plt.ylabel("Count")
plt.show()
```

```python
sns.scatterplot(data=df, x='Systolic', y='Diastolic', hue='Gender')
plt.title("Systolic vs Diastolic by Gender")
plt.xlabel("Systolic")
plt.ylabel("Diastolic")
plt.show()
```

```python
sns.boxplot(data=df, x='Diagnosis', y='Systolic')
plt.title("Systolic BP by Diagnosis")
plt.xticks(rotation=45)
plt.show()
```
